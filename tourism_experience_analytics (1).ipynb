## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# 1. Fix missing CityId in User table
user['CityId'] = user['CityId'].fillna(0).astype(int)

# 2. Fix missing CityName in City table
city['CityName'] = city['CityName'].fillna('Unknown')

# 3. Replace '-' placeholders with 'Unspecified' for readability
city['CityName']       = city['CityName'].replace('-', 'Unspecified')
country['Country']     = country['Country'].replace('-', 'Unspecified')
region['Region']       = region['Region'].replace('-', 'Unspecified')
continent['Continent'] = continent['Continent'].replace('-', 'Unspecified')
mode['VisitMode']      = mode['VisitMode'].replace('-', 'Unspecified')

# 4. Merge — only bring in the NAME column from each lookup table, not its ID columns
user_full = user.merge(city[['CityId', 'CityName']], on='CityId', how='left')
user_full = user_full.merge(country[['CountryId', 'Country']], on='CountryId', how='left')
user_full = user_full.merge(region[['RegionId', 'Region']], on='RegionId', how='left')
user_full = user_full.merge(continent[['ContinentId', 'Continent']], on='ContinentId', how='left')

# Attraction's type — item only has AttractionTypeId as its ID, no conflict here
item_full = item.merge(type_df, on='AttractionTypeId', how='left')

# Final master table
df = transaction.merge(user_full, on='UserId', how='left') \
                 .merge(item_full, on='AttractionId', how='left') \
                 .merge(mode, left_on='VisitMode', right_on='VisitModeId', how='left', suffixes=('', '_label'))

df.rename(columns={'VisitMode_label': 'VisitModeName'}, inplace=True)

print("Final shape:", df.shape)
df.head()
````

### What all manipulations have you done and insights you found?

Missing values were minimal and handled without deleting any rows: 4 users missing a CityId were assigned code 0 ("unspecified"), and 1 city missing a name was labeled "Unknown". Placeholder values ("-") across City, Country, Region, Continent, and Mode tables were relabeled "Unspecified" for readability. All 9 tables were then joined on their ID columns into a single master dataset (df), giving one row per transaction (52,930 rows total), enriched with the user's full geographic profile and the visited attraction's type and address. This master dataset is now ready for exploratory analysis and modeling.

### 3. Categorical Encoding

````python
# Encode your categorical columns
from sklearn.preprocessing import LabelEncoder

# Columns to encode for modeling
cat_cols = ['Continent', 'Region', 'Country', 'CityName', 'AttractionType', 'Attraction']

df_encoded = df.copy()
encoders = {}

for col in cat_cols:
    le = LabelEncoder()
    df_encoded[col + '_enc'] = le.fit_transform(df_encoded[col].astype(str))
    encoders[col] = le

print(df_encoded[[c + '_enc' for c in cat_cols]].head())
````

#### What all categorical encoding techniques have you used & why did you use those techniques?

Label Encoding was used (not One-Hot) because several columns have many unique categories (e.g., Country has 165, CityName has thousands) — one-hot encoding these would create thousands of extra columns and cause the dataset to explode in size.
